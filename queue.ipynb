# Queues can be implemented using linked list or with circular arrays

If we use normal array instead of a circular one, then one of the enque or deque operation will be O(n). Like if we make insertions to the end of the array, then dequeu happens at the front - then you have to move all the elements one index back to fill the hole.

But with a circular array, you keep a pointer called front, and when you dequeu, you deque at front (just set it to None or something) then move the front pointer forward (which can loop back).

```python
class ArrayQueue:
    def __resize(self):
        old = self._l
        self._l = [None] * (len(old) * 2)

        seek = self._front
        for k in range(self._size):
            self._l[k] = old[seek]
            seek = (seek + 1) % len(old)

        self._front = 0
        
    def __init__(self, size=8):
        self._size = 0
        self._front = 0
        self._l = [None]*size

    def __len__(self):
        return self._size

    def enque(self, v):
        if self._size == len(self._l):
            self.__resize()
        spot = (self._front + self._size) % len(self._l)
        self._l[spot] = v
        self._size += 1

    def dequeu(self):
        if self._size == 0:
            raise IndexError("empty queue")
        out = self._l[self._front]
        self._front = (self._front + 1) % len(self._l)
        self._size -= 1
        return out

    def __str__(self):
        return str(self._l)
```

```python
q = ArrayQueue()
print(len(q))
print(q)
try:
    q.dequeu()
except:
    print("exception on empty q deque")
q.enque(99)
print(len(q))
print(q)

for i in range(5):
    q.enque(i)

while True:
    try:
        print(q.dequeu())
    except:
        break
```
