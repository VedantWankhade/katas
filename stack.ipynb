# Stack
Stacks have constant time access to the top of the stack, and amortized (based  on dynamic array impolementation) insertion (push).

If we know the maximum size the stack can reach then we can avoid the amortization by specifying the total capacity in advance.

I dont do that in the below implementation as i am using just teh list class as base.

```python
class EmptyStackException(Exception):
    def __init__(self, msg):
        super().__init__(msg)
    
class ArrayStack():
    def __init__(self, cap=8):
        self._data = []

    def __len__(self):
        return len(self._data)

    def pop(self):
        if self._data:
            return self._data.pop()
        else:
            raise EmptyStackException("stack is empty vro")

    def top(self):
        if self._data:
            return self._data[-1]
    
    def push(self, v):
        self._data.append(v)

    def __str__(self):
        return str(self._data)

    def __empty__(self):
        return len(self) == 0
```

```python
s = ArrayStack()
print(s)
print(len(s))
try:
    s.pop()
except EmptyStackException as e:
    print("exception raised:", e)

s.push(1)
print(s)
print(len(s))
s.push(2)
s.push(3)
print(s)
print(len(s))

try:
    s.pop()
except EmptyStackException as e:
    print("exception raised:", e)
try:
    s.pop()
except EmptyStackException as e:
    print("exception raised:", e)
try:
    s.pop()
except EmptyStackException as e:
    print("exception raised:", e)
try:
    s.pop()
except EmptyStackException as e:
    print("exception raised:", e)
```

## Reverse data using stack

```python
from collections import deque # im just using deque as a stack here
def reverse(l:list)->list:
    stack = deque(l) # similar to stack.push(v) for v in list
    return [stack.pop() for _ in range(len(stack))]

l = [x for x in range(5)]
print(l)
lr = reverse(l)
print(lr)
```
