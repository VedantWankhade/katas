# Recursion
When one invocation of the function make a recursive call, that invocation is suspended until the recursive call completes

## Factorial
$$
n! = 
\begin{cases} 
1 & \text{if } n = 0 \\ 
n \times (n-1)! & \text{if } n > 0 
\end{cases}
$$


### Iterative approach

```python
def factorial_iter(n):
    out = 1
    for m in range(1, n+1):
        out *= m
    return out
```

```python
print(factorial_iter(0))
print(factorial_iter(1))
print(factorial_iter(2))
print(factorial_iter(5))
```

### Recursive approach

```python
factorial_rec = lambda n: n * factorial_rec(n-1) if n > 0 else 1
```

```python
print(factorial_rec(0))
print(factorial_rec(1))
print(factorial_rec(2))
print(factorial_rec(5))
```

## Binary Search
Two requirements for binary search:
1. The sequence should be indexable (i.e the elements of the sequence should be randomly accessible)
2. The sequence should be in sorted order

Disadvantages:
1. There is no guarantee which occurance (element or its index) will be returned if there are multiple occurances of target (key) in the sequence. (though it is possible to write modified version of binary search that retruns first or last occurance or even element >= target or for other cases)

```python
def binary_search(target: int, seq: list[int])->int:
    return _binary_search(target, seq, 0, len(seq)-1)

def _binary_search(target: int, seq: list[int], lo:int, hi:int)->int:
    if lo > hi:
        return -1
    
    mid = (hi + lo) // 2
    if seq[mid] == target:
        return mid
    elif seq[mid] > target:
        return _binary_search(target, seq, lo, mid-1)
    else:
        return _binary_search(target, seq, mid+1, hi)
```

```python
print(binary_search(99, []))
print(binary_search(99, [1]))
print(binary_search(99, [99]))
print(binary_search(99, [1, 99]))
print(binary_search(99, [1,2,3,4,5]))
print(binary_search(99, [1,2,3,4,5,99]))
```

In general, we can use the stack data structure to convert a recursive algorithm into a nonrecursive algorithm by managing the nesting of the recursive structure ourselves, rather than relying on the interpreter to do so.

Although this only shifts the memory usage from the interpreter to our stack, we may be able to reduce the memory usage by storing only the minimal information necessary.

Even better, some forms of recursion can be eliminated without any use of axillary memory. A notable such form is known as tail recursion. A recursion is a tail recursion if any recursive call that is made from one context is the very last operation in that context, with the return value of the recursive call (if any) immediately returned by the enclosing recursion. By necessity, a tail recursion must be a linear recursion (since there is no way to make a second recursive call if you must immediately return the result of the first).

## Linear recursion example
### Recursion sum

```python
# iterative approach
def sum_iter(seq: list[int]) -> int:
    sum = 0
    for n in seq:
        sum += n
    return sum

# recursive approach
def sum_rec(seq: list[int], index = 0):
    if index >= len(seq):
        return 0
    return seq[index] + sum_rec(seq, index+1)

print(sum_iter([]))
print(sum_iter([1]))
print(sum_iter([1, 2]))
print(sum_iter([1, 99]))

print(sum_rec([]))
print(sum_rec([1]))
print(sum_rec([1, 2]))
print(sum_rec([1, 99]))
```

### Recursive reversal
