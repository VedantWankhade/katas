# Everything in python is an object

Everything that is runtime entity is an object; even code.
Language syntax like keywords, operators etc are not.

## Some examples of objects:
1. ints
2. float
3. str
4. dict
5. list
6. functions
7. class
8. type
9. modules
10. code

Every object has an identity (like memory address) that is constant throughout lifetime of that object, type and value.

Ref: 
1. [Python Docs - Data Model](https://docs.python.org/3/reference/datamodel.html#data-model)

```python
num1 = 99
print(id(num1))
print(type(num1))

num2 = 99
print(id(num2))

print(id(num1) == id(num2)) # object reused

print(id(222)) # even without an identifier, object is created, has id, type and value
print(type(222))

def somefunc():
    pass

# functions are also objects
print(id(somefunc))
print(type(somefunc))
```

# type and object

`type` is a class actually a metaclass. While a class defines how instances should behave, metaclass defines how classes should behave.

`5` is an instance of `int` class, and `int` class is an instance of `type` metaclass.

`type` class is an instance of itself. Hence `type(type)` is `<class 'type'>`

`object` is the ultimate base class.

`type` is also a subclass of `object`.

Use `issubclass(MyClass, class)` to test if MyClass is a subclass of some class.
and `isinstance(objectInstance, ClassOrType)` to test if objectInstance is a instance of some class or type. 

Since `object` the class is an instance of `type` the class (aka type), and `type` the class is a subclass of `object` the class, it might seem like chicken age problem, and it is. So neither of them is declared before the other, the python interpreter declares them together somehow as part of the bootstrap process (in other words - dont worry about it). 

We can say `int` class (or any other class) extend the `object` class and is an instance of `type` class.

Ref:
1. [Python Basics](https://python-basics-tutorial.readthedocs.io/en/24.3.0/oop/types.html)
2. [YouTube - Metaclasses](https://www.youtube.com/watch?v=NAQEj-c2CI8)

```python
help(type)
help(object)
```

```python
print(type(int))
print(type(type))

print(isinstance(5, int))
class MyInt(int):
    pass

print(issubclass(MyInt, int))
print(issubclass(MyInt, object))
num = MyInt(5)
print(type(num))
print(type(MyInt))

print(type(object))

print(issubclass(type, object))

print(type.__bases__[0].__name__)

print(isinstance(int, type))
print(issubclass(int, type))
print(int.__class__)

a = 99
print(type(a))
print(isinstance(a, int))
print(isinstance(a, object))
print(isinstance(a, type))
print(a.__class__)
```

# Copying lists
## Shalow copy
1. list() constructor
2. [i:j:k] list slicing
3. list.copy() method on list object


```python
l1 = [1, 2, 3, 4]
l2 = l1
print(l1 is l2)
print(id(l1),id(l2))

l3 = l1.copy()
print(l3 is l1)

l4 = list(l1)
print(l4 is l1)

l5 = l1[:]
print(l5 is l1)
```

## Deep copy
1. copy.deepcopy() function fomr copy module

```python
l1 = [[1, 2], [3, 4]]
l2 = l1.copy()

print(l2 is l1)
# but
print(l2[0] is l1[0])

import copy
l3 = copy.deepcopy(l1)
print(l3 is l1)
# and
print(l3[0] is l1[0])
```

### Common operations on sequences:
1. Concatenating
2. Checking membership
3. Splicing
4. Replacing slices
5. Deleting element and slices
6. Lexicographical comparison

```python
l1 = [1,2,3]
print(1 in l1)
print(99 in l1)

d1 = {1: 'a'}
print(1 in d1)
print('a' in d1)

l1 = l1 + [99, 100]
print(l1)

l1 = l1 * 5
print(l1)

l2 = [1, 2, 3, 4]
print(l2[-1])
print(l2[-2])
print(l2[1:3])
print(l2[::2])
print(l2[::-1])
print(l2[3:1:-1])

print(l2)
l2[0:2] = [99, 100]
print(l2)
del l2[1]
print(l2)
del l2[1:]
print(l2)

l3 = [1, 2, 3, 4]
l4 = [1, 2, 3, 4]
print(l3 is l4)
print(l3 == l4)

l5 = [4, 1]
print(l3 < l5)
```

> [!WARNING]
> Important note about += operator with lists

Lists are mutable but when you concatenate list with +, it creates a new list and assignes back to the original identifier.
But the behaviour is different in case of += operator. It modifies the original list.

```python
l1 = [1, 2]
print(id(l1))
l1 = l1 + [99]
print(id(l1))

l1 += [100]
print(id(l1))
```
