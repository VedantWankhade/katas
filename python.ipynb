# type and object

`type` is a class actually a metaclass. While a class defines how instances should behave, metaclass defines how classes should behave.

`5` is an instance of `int` class, and `int` class is an instance of `type` metaclass.

`type` class is an instance of itself. Hence `type(type)` is `<class 'type'>`

`object` is the ultimate base class.

`type` is also a subclass of `object`.

Use `issubclass(MyClass, class)` to test if MyClass is a subclass of some class.
and `isinstance(objectInstance, ClassOrType)` to test if objectInstance is a instance of some class or type. 

Since `object` the class is an instance of `type` the class (aka type), and `type` the class is a subclass of `object` the class, it might seem like chicken age problem, and it is. So neither of them is declared before the other, the python interpreter declares them together somehow as part of the bootstrap process (in other words - dont worry about it). 

We can say `int` class (or any other class) extend the `object` class and is an instance of `type` class.

Ref:
1. [Python Basics](https://python-basics-tutorial.readthedocs.io/en/24.3.0/oop/types.html)
2. [YouTube - Metaclasses](https://www.youtube.com/watch?v=NAQEj-c2CI8)

````python
help(type)
help(object)
````

````python
print(type(int))
print(type(type))

print(isinstance(5, int))
class MyInt(int):
    pass

print(issubclass(MyInt, int))
print(issubclass(MyInt, object))
num = MyInt(5)
print(type(num))
print(type(MyInt))

print(type(object))

print(issubclass(type, object))

print(type.__bases__[0].__name__)

print(isinstance(int, type))
print(issubclass(int, type))
print(int.__class__)

a = 99
print(type(a))
print(isinstance(a, int))
print(isinstance(a, object))
print(isinstance(a, type))
print(a.__class__)
````

> Important note about += operator with lists

Lists are mutable but when you concatenate list with +, it creates a new list and assignes back to the original identifier.
But the behaviour is different in case of += operator. It modifies the original list.

````python
l1 = [1, 2]
print(id(l1))
l1 = l1 + [99]
print(id(l1))

l1 += [100]
print(id(l1))
````

# Function parameters

## Argument passing
Python uses pass by reference. Also called as pass by assignment in python.

The arguments are literally assigned to the parameters. Hence aliased.

So mutable objects modified inside a function will actually modify the global instance. 

````python
def modify_mutable(l):
    print("inside function:", id(l))
    l.append(99) # global instance modified
    print("inside function after modification:", id(l))

l = [1, 2, 3]
print(l)
print(id(l))
modify_mutable(l)
print(l)
print(id(l))

def modify_immutable(a):
    print("inside function:", id(a))
    a = 99 # new object assigned to local variable a
    print("inside function after modification:", id(a)) 
    
a = 2
print(a)
print(id(a))
modify_immutable(2)
print(a)
print(id(a))
````

## Default values for parameters
You can give the parameters of a function default values and making them optional for the caller to pass.
The default parameters should be declared last in the parameter list.

````python
def add_by(num, by = 10):
    return num + by

print(add_by(2))
print(add_by(2, 20))

# def somefunc(by = 22, num)) # wrong declaration
````

## Positional and keyword arguments
You can pass the arguments by position or by keyword.

You can mix them too, but keyword arguments need to come after positional arguments.

The author of the function can also restrict the arguments to be passed by keyword with *. Its quite useful to make APIs backward compatible

Similarly you can enforce position arguments with /.
Arguments preceding / need to be positional only: 
```py
def function_name(positional_only, /, positional_or_keyword, *, keyword_only):
    pass
```


````python
def sub_by(num, by):
    return num - by

print(sub_by(num = 2, by = 1))
print(sub_by(2, by = 1))

# these will throw error
# print(sub_by(2, num = 2)) # this passes value to num twice
# print(sub_by(num = 2, 1)) 

def mul_by(num, *, by): # all parameters after * need to be passed with keyword
    return num * by

print(mul_by(2, by=22))
# these will throw error
# print(mul_by(2, 3))
````

# Lazy iterators
Values in iterators are lazily evaluated.

We see lazy evaluation used in many of Python’s libraries. For example, the dictionary class supports methods keys(), values(), and items(), which respectively produce a “view” of all keys, values, or (key,value) pairs within a dictionary. 

None of these methods produces an explicit list of results. Instead, the views that are produced are iterable objects based upon the actual contents of the dictionary. An explicit list of values from such an iteration can be immediately constructed by calling the list class constructor with the iteration as a parameter. For example, the syntax list(range(1000)) produces a list instance with values from 0 to 999, while the syntax list(d.values()) produces a list that has elements based upon the current values of dictionary d. We can similarly construct a tuple or set instance based upon a given iterable.

`factors = [k for k in range(1,n+1) if n % k = = 0]` Python supports similar comprehension syntaxes that respectively produce a set, generator, or dictionary. We compare those syntaxes using our example for producing the squares of numbers.

The generator syntax is particularly attractive when results do not need to be stored in memory. For example, to compute the sum of the first n squares, the generator syntax, `total = sum(k*k for k in range(1, n+1))`, is preferred to the use of an explicitly instantiated list comprehension as the parameter.

````python
square_generator = (x * x for x in range(1, 21))
print(square_generator)
print(list(square_generator))
````

# Random module
1. `random()` returns random number between 0.0 to 1.0
2. `randint(a, b)` random int between a and b both inclusive
3. `randrange(start, stop)` random int range between start, stop, upper limit exclusive
4. `choice(seq)` select a random element from sequence
5. `sample(seq, k=n)` select multiple random uniqu elements

````python
import random 

print(random.random())
print(random.random())
print(random.randint(2, 5))
print(random.randint(2, 5))
print(random.randrange(2, 6))
print(random.randrange(2, 6))
print(random.choice([1, 2, 3]))
print(random.choice([1, 2, 3]))
print(random.sample([1, 2, 3, 4, 5], 3))
print(random.sample([1, 2, 3, 4, 5], 3))
      
````
